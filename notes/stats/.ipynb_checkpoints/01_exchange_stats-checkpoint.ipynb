# Exchange Monthly Stats

Goal:

1. Measure how many new users each exchange gets per month in recent months.
2. Measure how much volume each exchange gets per month.
3. If possible, compute volume per user per month.

The API requests are split into calendar-month windows so each request stays within the 1-month limit. The exchange list, date range, and metric list are all configurable.

```python
from pathlib import Path
import sys

import pandas as pd

sys.path.insert(0, str(Path('../../script').resolve()))

from exchange_stats import (  # noqa: E402
    DEFAULT_METRICS,
    build_monthly_summary,
    fetch_explorer_data_for_range,
    iter_month_windows,
)
```

```python
# Configurable inputs
exchange_slugs = [
    'hyperliquid',
    'lighter',
    'extended',
    'variational',
    'nado',
    'pacifica',
]

from_date = '2025-01-01'
to_date = '2025-06-30'
metrics = DEFAULT_METRICS

list(iter_month_windows(pd.Timestamp(from_date).date(), pd.Timestamp(to_date).date()))[:3]
```

## Fetch data

The script uses the same request headers as the curl example in `docs/volume_stats.md` and sends one request per month window.

```python
raw_df = fetch_explorer_data_for_range(
    slugs=exchange_slugs,
    metrics=metrics,
    from_date=from_date,
    to_date=to_date,
)

raw_df.head()
```

## Monthly summary

For the main questions, the important outputs are:

- `new_users`: monthly total new users
- `volume`: monthly total volume
- `volume_per_user`: `volume / new_users`


```python
summary_df = build_monthly_summary(raw_df)
summary_df
```

```python
# Optional: save outputs for later analysis
output_dir = Path('../../data/exchange_stats').resolve()
output_dir.mkdir(parents=True, exist_ok=True)

raw_df.to_csv(output_dir / 'raw_rows.csv', index=False)
summary_df.to_csv(output_dir / 'monthly_summary.csv', index=False)

output_dir
```

## Notes

- If `new_users` is zero for a month, `volume_per_user` is left blank to avoid division by zero.
- `tvl` and `open_interest` are treated as end-of-month snapshots.
- If the API adds new metrics later, the script can fetch them by extending the `metrics` list.
