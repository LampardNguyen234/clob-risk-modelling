# Token Listing Evaluation — RISEx

Operationalises the token listing framework from `docs/Token_Listing.md`.

**Pipeline:**
1. Load token list from `config.yml`
2. Fetch and cache four API responses per token (`script/token_listing_fetcher.py`)
3. Evaluate hard gates (FDMC, exchange score, safety, sanctions, feed grade)
4. Compute six market score metrics for gate-passing tokens
5. Assign destination tier (score → feed grade ceiling → Prime whitelist)
6. Surface soft flags and recommend preListing duration

Set `REFRESH = True` to bypass the local cache and re-fetch all APIs.

```python
import sys
import warnings
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
import pandas as pd
from IPython.display import display

sys.path.insert(0, str(Path.cwd().parent.parent / "script"))
import token_listing_fetcher as fetcher

pd.set_option("display.max_colwidth", None)
pd.set_option("display.float_format", "{:,.4f}".format)

CONFIG_PATH = Path("config.yml")
REFRESH = False  # set True to bypass cache and re-fetch all APIs

# Exchange tier classification (Token_Listing.md §Exchange Score)
TIER1_SLUGS = {"binance", "coinbase", "okx", "bybit", "kraken"}
TIER2_SLUGS = {"kucoin", "gate-io", "bitstamp", "upbit"}
USD_QUOTES  = {"USDT", "USDC", "USD"}

# Market score weights and reference values (Token_Listing.md §Market Score)
METRIC_WEIGHT = {1: 25, 2: 20, 3: 15, 4: 20, 5: 15, 6: 5}
METRIC_REF    = {1: 5_000_000, 2: 0.30, 3: 0.25, 4: 500_000_000, 5: 0.85, 6: 0.0005}
METRIC_INV    = {2, 3, 6}  # inverted: lower raw value → higher score

PRIME_WHITELIST = {"BTC", "ETH"}
SCORE_TIERS = [(85, "Prime"), (70, "Major"), (50, "Mid-Cap"),
               (30, "Small-Cap"), (10, "Micro-Cap"), (0, "Rejected")]
FEED_CEILING  = {"A": "Prime", "B": "Major", "C": "Small-Cap", "D": "Micro-Cap"}
TIER_ORDER    = ["Prime", "Major", "Mid-Cap", "Small-Cap", "Micro-Cap", "Rejected"]
TIER_PARAMS   = {
    "Prime":     dict(max_lev=25, imr="4.00%",  mmr="2.67%",  cmr="1.78%",  oi_lower="$50M",  oi_upper="$120M"),
    "Major":     dict(max_lev=20, imr="5.00%",  mmr="3.33%",  cmr="2.22%",  oi_lower="$10M",  oi_upper="$20M"),
    "Mid-Cap":   dict(max_lev=10, imr="10.00%", mmr="6.67%",  cmr="4.44%",  oi_lower="$2M",   oi_upper="$10M"),
    "Small-Cap": dict(max_lev=5,  imr="20.00%", mmr="13.33%", cmr="8.89%",  oi_lower="$500K", oi_upper="$3M"),
    "Micro-Cap": dict(max_lev=3,  imr="33.33%", mmr="22.22%", cmr="14.81%", oi_lower="$200K", oi_upper="$1M"),
    "Rejected":  dict(max_lev="—", imr="—",     mmr="—",      cmr="—",      oi_lower="—",     oi_upper="—"),
}

def fmt_usd(v):
    if v >= 1e9:  return f"${v/1e9:.2f}B"
    if v >= 1e6:  return f"${v/1e6:.2f}M"
    if v >= 1e3:  return f"${v/1e3:.1f}K"
    return f"${v:.2f}"
```

```python
# Load config and fetch all data (cache-first)
tokens = fetcher.load_config(CONFIG_PATH)
raw = {}
for token in tokens:
    sym = token["symbol"]
    print(f"Fetching {sym}...")
    raw[sym] = fetcher.fetch_all(token, refresh=REFRESH)
print("\nDone.")
```

---
## Hard Gates

All gates must pass before scoring proceeds. A single failure is a hard reject.

| Gate | Requirement |
|---|---|
| FDMC | ≥ $50M fully diluted market cap |
| Exchange score | ≥ 5 weighted points **and** ≥ 1 Tier-1 exchange |
| Safety score | ≥ 60 (manual review; 60–74 = conditional pass + 14d preListing) |
| Sanctions | Not on OFAC or equivalent |
| Feed grade | ≥ C (Stork publisher count ≥ 2); assumed Grade A for this run |

Exchange score filters out pairs where `outlierDetected=1`, `volumeExcluded=1`, or `marketReputation < 0.1`, and only counts USD-quoted pairs (USDT/USDC/USD).

```python
def compute_exchange_gate(market_pairs_data: dict) -> dict:
    """Filter, deduplicate, and score exchanges from CMC market-pairs response."""
    per_exchange: dict[str, dict] = {}
    for pair in market_pairs_data.get("data", {}).get("marketPairs", []):
        if pair.get("quoteSymbol") not in USD_QUOTES:
            continue
        if pair.get("outlierDetected") or pair.get("volumeExcluded"):
            continue
        if (pair.get("marketReputation") or 0) < 0.1:
            continue
        slug = pair["exchangeSlug"]
        if slug not in per_exchange:
            per_exchange[slug] = {"volume": 0.0, "depth_bid": 0.0, "depth_ask": 0.0}
        per_exchange[slug]["volume"]    += pair.get("volumeUsd", 0.0)
        per_exchange[slug]["depth_bid"] += pair.get("depthUsdNegativeTwo", 0.0)
        per_exchange[slug]["depth_ask"] += pair.get("depthUsdPositiveTwo", 0.0)

    def tier_weight(slug):
        if slug in TIER1_SLUGS: return 3
        if slug in TIER2_SLUGS: return 2
        return 1

    score     = sum(tier_weight(s) for s in per_exchange)
    has_tier1 = any(s in TIER1_SLUGS for s in per_exchange)
    return {
        "score":        score,
        "has_tier1":    has_tier1,
        "pass":         score >= 5 and has_tier1,
        "per_exchange": per_exchange,
    }


gates = {}
for token in tokens:
    sym       = token["symbol"]
    d         = raw[sym]
    overrides = token.get("overrides", {})
    stats     = d["cmc_detail"].get("data", {}).get("statistics", {})

    fdmc = overrides.get(
        "fullyDilutedMarketCap",
        stats.get("fullyDilutedMarketCap") or stats.get("marketCap", 0)
    )

    ex         = compute_exchange_gate(d["cmc_market_pairs"])
    safety     = token["safety_score"]
    sanctions  = token["sanctions_pass"]
    feed_grade = token.get("feed_grade", "A")

    gates[sym] = {
        "fdmc":               fdmc,
        "fdmc_pass":          fdmc >= 50_000_000,
        "exchange_score":     ex["score"],
        "exchange_has_tier1": ex["has_tier1"],
        "exchange_pass":      ex["pass"],
        "safety_score":       safety,
        "safety_pass":        safety >= 60,
        "safety_conditional": 60 <= safety < 75,
        "sanctions_pass":     sanctions,
        "feed_grade":         feed_grade,
        "feed_pass":          feed_grade not in {"F"},
        "per_exchange":       ex["per_exchange"],
        "all_pass":           fdmc >= 50_000_000 and ex["pass"]
                              and safety >= 60 and sanctions and feed_grade not in {"F"},
    }
```

### Hard Gate Results

```python
def gate_label(passed, conditional=False):
    if conditional: return "PASS*"
    return "PASS" if passed else "FAIL"

rows = []
for sym, g in gates.items():
    rows.append({
        "Token":          sym,
        "FDMC":           fmt_usd(g["fdmc"]),
        "FDMC gate":      gate_label(g["fdmc_pass"]),
        "Exch score":     g["exchange_score"],
        "Tier-1 present": "YES" if g["exchange_has_tier1"] else "NO",
        "Exch gate":      gate_label(g["exchange_pass"]),
        "Safety score":   g["safety_score"],
        "Safety gate":    gate_label(g["safety_pass"], g["safety_conditional"]),
        "Sanctions":      gate_label(g["sanctions_pass"]),
        "Feed grade":     g["feed_grade"],
        "Feed gate":      gate_label(g["feed_pass"]) + " (assumed A)",
        "OVERALL":        gate_label(g["all_pass"]),
    })

gate_df = pd.DataFrame(rows).set_index("Token")
gate_cols = ["FDMC gate", "Exch gate", "Safety gate", "Sanctions", "Feed gate", "OVERALL"]

def colour_gate(val):
    if str(val).startswith("PASS*"): return "color: orange; font-weight: bold"
    if str(val).startswith("PASS"):  return "color: green; font-weight: bold"
    if str(val).startswith("FAIL"):  return "color: red; font-weight: bold"
    return ""

styled = gate_df.style.applymap(colour_gate, subset=gate_cols)
display(styled)

# Annotate conditional passes
conditional = [s for s, g in gates.items() if g["safety_conditional"]]
if conditional:
    print(f"\nPASS* (conditional): {', '.join(conditional)} — safety score 60–74; "
          "14-day preListing mandatory; escalate to risk team.")
failed = [s for s, g in gates.items() if not g["all_pass"]]
if failed:
    print(f"Hard rejected (skipped in scoring): {', '.join(failed)}")
```

## Market Score

Six metrics normalised linearly against reference values, summed to $S \in [0, 100]$.

$$s_i = \min\!\left(\frac{v_i}{r_i},\; 1\right) \times w_i \quad \text{(standard)} \qquad s_i = \min\!\left(\frac{r_i}{v_i},\; 1\right) \times w_i \quad \text{(inverted)}$$

| # | Metric | Weight | Reference | Direction |
|---|---|---|---|---|
| 1 | Spot depth | 25 | $5M | Standard |
| 2 | Realised volatility | 20 | 30% ann. | Inverted |
| 3 | Max drawdown | 15 | 25% | Inverted |
| 4 | Spot volume | 20 | $500M 30d avg daily | Standard |
| 5 | Volume concentration | 15 | 0.85 | Standard |
| 6 | Book spread | 5 | 0.05% | Inverted |

```python
from datetime import date

TODAY_UTC = datetime.now(timezone.utc).replace(hour=0, minute=0, second=0, microsecond=0)


def _parse_quotes(historical_data: dict) -> list[dict]:
    """Extract complete daily candles (timeClose < today 00:00 UTC)."""
    quotes = historical_data.get("data", {}).get("quotes", [])
    complete = []
    for q in quotes:
        tc = q.get("timeClose", "")
        if tc:
            try:
                close_dt = datetime.fromisoformat(tc.replace("Z", "+00:00"))
                if close_dt < TODAY_UTC:
                    complete.append(q)
            except ValueError:
                pass
    return complete


def compute_realised_vol(historical_data: dict) -> tuple[float, int, str | None]:
    """
    Annualised realised volatility using sqrt(365) * std(log_returns).
    Uses trailing 30 complete daily closing prices (or fewer if history is short).
    Returns (sigma_ann, n_days_used, flag_or_None).
    """
    quotes = _parse_quotes(historical_data)[-30:]  # trailing 30
    n = len(quotes)
    if n < 7:
        return 0.0, n, f"insufficient history ({n}d < 7d); s2 scored 0"
    closes = np.array([q["quote"]["close"] for q in quotes], dtype=float)
    log_ret = np.diff(np.log(closes))
    sigma = float(np.sqrt(365) * np.std(log_ret, ddof=1))
    flag = f"short history ({n}d < 30d)" if n < 30 else None
    return sigma, n, flag


def compute_max_dd(historical_data: dict) -> tuple[float, int, str | None]:
    """
    Maximum drawdown over trailing 30 candles.
    Running peak = daily high; trough = daily low.
    Liquidations are triggered intraday, so close-to-close DD understates risk.
    Returns (max_dd, n_days_used, flag_or_None).
    """
    quotes = _parse_quotes(historical_data)[-30:]
    n = len(quotes)
    if n < 7:
        return 0.0, n, f"insufficient history ({n}d < 7d); s3 scored 0"
    peak = 0.0
    max_dd = 0.0
    for q in quotes:
        high = q["quote"]["high"]
        low  = q["quote"]["low"]
        peak = max(peak, high)
        if peak > 0:
            max_dd = max(max_dd, (peak - low) / peak)
    flag = f"short history ({n}d < 30d)" if n < 30 else None
    return float(max_dd), n, flag


def compute_spot_depth(per_exchange: dict) -> float:
    """Sum min(bid_depth, ask_depth) across top 10 exchanges by volume."""
    top10 = sorted(per_exchange.items(), key=lambda kv: kv[1]["volume"], reverse=True)[:10]
    return sum(min(v["depth_bid"], v["depth_ask"]) for _, v in top10)


def compute_hhi_concentration(per_exchange: dict) -> float:
    """1 - HHI over per-exchange volumes. Returns concentration in [0, 1]."""
    total = sum(v["volume"] for v in per_exchange.values())
    if total == 0:
        return 0.0
    hhi = sum((v["volume"] / total) ** 2 for v in per_exchange.values())
    return 1.0 - hhi


def compute_book_spread(cg_tickers: dict, top10_slugs: list[str]) -> tuple[float, str | None]:
    """
    Mean bid_ask_spread_percentage across top-10 exchanges.
    Matches CMC exchangeSlug to CoinGecko market.identifier.
    Excludes anomaly/stale tickers; only USD-quoted pairs.
    Returns (mean_spread_fraction, flag_or_None).
    """
    tickers = cg_tickers.get("tickers", [])
    if not tickers:
        return 0.0, "CG_NO_DATA: no tickers returned; spread scored 0"

    slug_set = set(top10_slugs)
    # Aggregate spread per exchange (average across USD pairs on that exchange)
    per_exchange: dict[str, list[float]] = {}
    for t in tickers:
        slug = t.get("market", {}).get("identifier", "")
        if slug not in slug_set:
            continue
        if t.get("target") not in USD_QUOTES:
            continue
        if t.get("is_anomaly") or t.get("is_stale"):
            continue
        spread_pct = t.get("bid_ask_spread_percentage")
        if spread_pct is None or spread_pct <= 0:
            continue
        per_exchange.setdefault(slug, []).append(spread_pct)

    if not per_exchange:
        return 0.0, "CG_NO_MATCH: no top-10 exchanges matched in CoinGecko tickers; spread scored 0"

    exchange_means = [np.mean(spreads) for spreads in per_exchange.values()]
    mean_spread_pct = float(np.mean(exchange_means))
    # Convert from percentage (e.g. 0.010013) to fraction (0.00010013)
    return mean_spread_pct / 100.0, None


def score_metric(v: float, ref: float, w: int, inverted: bool) -> float:
    if v <= 0:
        return 0.0
    ratio = ref / v if inverted else v / ref
    return min(ratio, 1.0) * w
```

```python
scores: dict[str, dict] = {}
flags:  dict[str, list] = {}  # sym -> [(metric_id, description), ...]

for token in tokens:
    sym = token["symbol"]
    if not gates[sym]["all_pass"]:
        continue

    d            = raw[sym]
    overrides    = token.get("overrides", {})
    per_exchange = gates[sym]["per_exchange"]
    token_flags: list[tuple[str, str]] = []

    # Metric 1: Spot depth
    m1_val = compute_spot_depth(per_exchange)
    s1 = score_metric(m1_val, METRIC_REF[1], METRIC_WEIGHT[1], False)

    # Metric 2: Realised volatility
    sigma, n_v, flag2 = compute_realised_vol(d["cmc_historical"])
    if flag2: token_flags.append(("m2_vol", flag2))
    s2 = score_metric(sigma, METRIC_REF[2], METRIC_WEIGHT[2], True)

    # Metric 3: Max drawdown
    max_dd, n_d, flag3 = compute_max_dd(d["cmc_historical"])
    if flag3: token_flags.append(("m3_dd", flag3))
    s3 = score_metric(max_dd, METRIC_REF[3], METRIC_WEIGHT[3], True)

    # Metric 4: Spot volume (30d avg daily)
    stats = d["cmc_detail"].get("data", {}).get("statistics", {})
    volume30d = overrides.get("volume30d", stats.get("volume30d", 0) or 0)
    avg_daily = volume30d / 30
    s4 = score_metric(avg_daily, METRIC_REF[4], METRIC_WEIGHT[4], False)

    # Metric 5: Volume concentration (1 - HHI)
    conc = compute_hhi_concentration(per_exchange)
    s5 = score_metric(conc, METRIC_REF[5], METRIC_WEIGHT[5], False)

    # Metric 6: Book spread (CoinGecko)
    top10_slugs = [k for k, _ in sorted(
        per_exchange.items(), key=lambda kv: kv[1]["volume"], reverse=True
    )[:10]]
    spread, flag6 = compute_book_spread(d["coingecko_tickers"], top10_slugs)
    if flag6: token_flags.append(("m6_spread", flag6))
    s6 = score_metric(spread, METRIC_REF[6], METRIC_WEIGHT[6], True)

    total = s1 + s2 + s3 + s4 + s5 + s6
    scores[sym] = {
        "s1": s1, "v1": m1_val,
        "s2": s2, "v2": sigma,   "n_v": n_v,
        "s3": s3, "v3": max_dd,  "n_d": n_d,
        "s4": s4, "v4": avg_daily,
        "s5": s5, "v5_hhi": 1 - conc,
        "s6": s6, "v6": spread,
        "total": total,
    }
    flags[sym] = token_flags
```

### Metric Scores

```python
rows = []
for sym, sc in scores.items():
    rows.append({
        "Token":             sym,
        "s1 Depth (25)":     f"{sc['s1']:.1f}  [{fmt_usd(sc['v1'])}]",
        "s2 Vol (20)":       f"{sc['s2']:.1f}  [{sc['v2']*100:.0f}% ann, {sc['n_v']}d]",
        "s3 MaxDD (15)":     f"{sc['s3']:.1f}  [{sc['v3']*100:.1f}%, {sc['n_d']}d]",
        "s4 Volume (20)":    f"{sc['s4']:.1f}  [{fmt_usd(sc['v4'])}/d]",
        "s5 Conc (15)":      f"{sc['s5']:.1f}  [HHI={sc['v5_hhi']:.3f}]",
        "s6 Spread (5)":     f"{sc['s6']:.1f}  [{sc['v6']*100:.4f}%]",
        "Total S":           f"{sc['total']:.1f}",
    })

score_df = pd.DataFrame(rows).set_index("Token")
display(score_df)
```

## Tier Assignment

Three steps (from `docs/Token_Listing.md §Tier Assignment`):

1. **Score → unconstrained tier**: S ≥ 85 → Prime, ≥ 70 → Major, ≥ 50 → Mid-Cap, ≥ 30 → Small-Cap, ≥ 10 → Micro-Cap, else → Rejected.
2. **Apply feed grade ceiling**: destination = min(score tier, grade ceiling).
3. **Prime whitelist**: only BTC and ETH are Prime-eligible; all others are demoted to Major if Step 2 yields Prime.

```python
def score_to_tier(s: float) -> str:
    for threshold, tier in SCORE_TIERS:
        if s >= threshold:
            return tier
    return "Rejected"


def apply_feed_ceiling(tier: str, ceiling: str) -> str:
    """Return the more conservative (lower) of tier and ceiling."""
    i_tier    = TIER_ORDER.index(tier)
    i_ceiling = TIER_ORDER.index(ceiling)
    return tier if i_tier >= i_ceiling else ceiling


def recommend_prelisting(tier: str, feed_grade: str, score: float,
                         safety_conditional: bool) -> str:
    if safety_conditional:
        return "14d (safety conditional)"
    if tier in {"Prime", "Major"} and feed_grade == "A" and score >= 70:
        return "24h"
    if tier in {"Mid-Cap", "Small-Cap"} or feed_grade == "B":
        return "7d"
    return "14d"


tiers: dict[str, dict] = {}
for sym, sc in scores.items():
    raw_tier = score_to_tier(sc["total"])
    ceiling  = FEED_CEILING.get(gates[sym]["feed_grade"], "Micro-Cap")
    capped   = apply_feed_ceiling(raw_tier, ceiling)
    final    = "Major" if capped == "Prime" and sym not in PRIME_WHITELIST else capped
    tiers[sym] = {
        "raw_tier":         raw_tier,
        "feed_ceiling":     ceiling,
        "capped_tier":      capped,
        "final_tier":       final,
        "whitelist_demoted": capped == "Prime" and sym not in PRIME_WHITELIST,
    }
```

## Summary

```python
summary_rows = []
for token in tokens:
    sym = token["symbol"]
    g   = gates[sym]

    if not g["all_pass"]:
        failed_gates = []
        if not g["fdmc_pass"]:       failed_gates.append("FDMC")
        if not g["exchange_pass"]:   failed_gates.append("Exchange")
        if not g["safety_pass"]:     failed_gates.append("Safety")
        if not g["sanctions_pass"]:  failed_gates.append("Sanctions")
        if not g["feed_pass"]:       failed_gates.append("Feed")
        summary_rows.append({
            "Token": sym, "Gate": "FAIL", "Score S": "—",
            "Raw Tier": "—", "Feed Grade": g["feed_grade"],
            "Destination Tier": f"REJECTED ({', '.join(failed_gates)})",
            "Max Lev": "—", "IMR": "—", "OI Lower": "—", "OI Upper": "—",
            "preListing": "—", "Soft Flags": "—",
        })
        continue

    sc   = scores[sym]
    tier = tiers[sym]
    p    = TIER_PARAMS[tier["final_tier"]]
    flag_strs = "; ".join(f"{ft}: {fd}" for ft, fd in flags[sym]) or "none"
    dest_label = tier["final_tier"]
    if tier["whitelist_demoted"]:
        dest_label += " [whitelist cap]"

    summary_rows.append({
        "Token":            sym,
        "Gate":             "PASS*" if g["safety_conditional"] else "PASS",
        "Score S":          f"{sc['total']:.1f}",
        "Raw Tier":         tier["raw_tier"],
        "Feed Grade":       g["feed_grade"],
        "Destination Tier": dest_label,
        "Max Lev":          f"{p['max_lev']}x",
        "IMR":              p["imr"],
        "OI Lower":         p["oi_lower"],
        "OI Upper":         p["oi_upper"],
        "preListing":       recommend_prelisting(
                                tier["final_tier"], g["feed_grade"],
                                sc["total"], g["safety_conditional"]),
        "Soft Flags":       flag_strs,
    })

summary_df = pd.DataFrame(summary_rows).set_index("Token")

def colour_summary(val):
    if str(val).startswith("PASS*"): return "color: orange; font-weight: bold"
    if str(val).startswith("PASS"):  return "color: green; font-weight: bold"
    if str(val).startswith("FAIL") or str(val).startswith("REJECT"): return "color: red; font-weight: bold"
    return ""

display(summary_df.style.applymap(colour_summary, subset=["Gate", "Destination Tier"]))
```

## Soft Flags

Non-blocking. Attached to the listing record for reviewer reference.

| Flag | Meaning |
|---|---|
| `m2_vol: short history` | Volatility computed on < 30d; estimate is noisier |
| `m2_vol / m3_dd: insufficient history` | < 7d data; both metrics scored 0 |
| `m6_spread: CG_NO_DATA` | CoinGecko returned no tickers; spread scored 0 (5 pts missing) |
| `m6_spread: CG_NO_MATCH` | No top-10 exchanges matched in CoinGecko response; spread scored 0 |

```python
any_flags = False
for sym, sym_flags in flags.items():
    if sym_flags:
        any_flags = True
        print(f"{sym}:")
        for ft, fd in sym_flags:
            print(f"  [{ft}] {fd}")
if not any_flags:
    print("No soft flags.")
```
